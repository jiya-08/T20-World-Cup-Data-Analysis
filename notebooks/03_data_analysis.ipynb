# 03. Statistical Data Analysis & Business Insights
## ICC Men's T20 World Cup 2022 Analytics

### Objective:
Conduct deep-dive statistical analysis on tournament KPIs, team standings, defending vs chasing dynamics, top batsmen, leading bowlers, and venue characteristics.


```python
import os
import sys
import pandas as pd
import numpy as np

sys.path.append(os.path.abspath('..'))
from src.analysis import (
    load_processed_data,
    get_tournament_kpis,
    get_batting_first_vs_chasing_analysis,
    get_top_run_scorers,
    get_top_wicket_takers,
    get_best_bowling_figures,
    get_highest_individual_scores,
    get_venue_performance
)

data = load_processed_data()
```

### 1. Tournament Overview KPIs


```python
kpis = get_tournament_kpis(data['matches'], data['batting'], data['bowling'])
for k, v in kpis.items():
    print(f'{k:<30}: {v}')
```

### 2. Batting First (Defending) vs Chasing Analysis


```python
toss_impact = get_batting_first_vs_chasing_analysis(data['matches'])
print(f"Defending Wins: {toss_impact['defending_wins']} ({toss_impact['defending_win_percentage']} %)")
print(f"Chasing Wins:   {toss_impact['chasing_wins']} ({toss_impact['chasing_win_percentage']} %)")
print('\nVenue-wise Win Breakdown:')
toss_impact['venue_win_breakdown']
```

### 3. Top 10 Run Scorers of the Tournament


```python
top_batsmen = get_top_run_scorers(data['batting'], limit=10)
top_batsmen[['player_name', 'team_innings', 'total_runs', 'batting_average', 'strike_rate', 'fours', 'sixes', 'highest_score']]
```

### 4. Top 10 Wicket Takers of the Tournament


```python
top_bowlers = get_top_wicket_takers(data['bowling'], limit=10)
top_bowlers[['player_name', 'bowling_team', 'wickets', 'bowling_average', 'economy_rate', 'dot_ball_pct', 'maidens']]
```

### 5. Best Individual Match Performances


```python
best_bowling = get_best_bowling_figures(data['bowling'], limit=5)
print('Best Bowling Figures:')
print(best_bowling[['player_name', 'bowling_team', 'figures', 'economy', 'dot_balls']])

highest_scores = get_highest_individual_scores(data['batting'], limit=5)
print('\nHighest Individual Batting Scores:')
print(highest_scores[['player_name', 'team_innings', 'runs', 'balls', 'strike_rate', 'dismissal_status']])
```
